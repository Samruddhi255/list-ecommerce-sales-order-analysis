```python

import pandas as pd

# Read the uploaded CSV files
orders = pd.read_csv("olist_orders_dataset.csv")
items = pd.read_csv("olist_order_items_dataset.csv")
products = pd.read_csv("olist_products_dataset.csv")
payments = pd.read_csv("olist_order_payments_dataset.csv")

# Combine orders with their items
data = orders.merge(items, on="order_id", how="inner")

# Add product information
data = data.merge(products, on="product_id", how="left")

# Calculate total payment for each order
payment_total = payments.groupby("order_id", as_index=False)[
    "payment_value"
].sum()

data = data.merge(payment_total, on="order_id", how="left")

# Convert the order date into a proper date format
data["order_purchase_timestamp"] = pd.to_datetime(
    data["order_purchase_timestamp"], errors="coerce"
)

# Check the combined dataset
print("Number of rows:", data.shape[0])
print("Number of columns:", data.shape[1])
print("\nColumn names:")
print(data.columns.tolist())

data.head()
```

```python

data.to_csv("olist_combined_dataset.csv", index=False)

from google.colab import files
files.download("olist_combined_dataset.csv")
```
